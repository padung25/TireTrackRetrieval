```python
from ultralytics import YOLO

model = YOLO("yolo11s.yaml") 

model.train(
    data="/kaggle/input/newdata/data_detect.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    optimizer='AdamW',
    lr0=0.002,
    momentum=0.9,
    weight_decay = 0.0005,
    dropout=0.2,
    name="tire_detection",
    pretrained=True
)
```

# Đánh giá

```python
from ultralytics import YOLO

# Load mô hình đã huấn luyện
model = YOLO("runs/detect/tire_detection/weights/best.pt")

# In ra bảng chỉ số
results = model.val()
print("mAP@0.5:", results.box.map50)
print("mAP@0.5:0.95:", results.box.map)
print("Precision:", results.box.p)
print("Recall:", results.box.r)
```
